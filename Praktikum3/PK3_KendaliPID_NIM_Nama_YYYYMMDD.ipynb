**1. Upload Semua File Sekaligus**

```python
from google.colab import files
uploaded = files.upload()
```

**2. Load & Simpan ke Dictionary**

```python
import pandas as pd

data = {}

for filename in uploaded.keys():
    df = pd.read_csv(filename)

    # waktu ke detik
    df['time_s'] = df['time_ms'] / 1000

    data[filename] = df

list(data.keys())
```

**PLOT 1 — Perbandingan ERROR**

```python
import matplotlib.pyplot as plt

plt.figure()

for name, df in data.items():
    plt.plot(df['time_s'], df['error'], label=name)

plt.xlabel('Time (s)')
plt.ylabel('Error')
plt.title('Perbandingan Error vs Time')
plt.legend()
plt.grid()

plt.show()
```

**PLOT 2 — Perbandingan CONTROL (u(t))**

```python
plt.figure()

for name, df in data.items():
    plt.plot(df['time_s'], df['control'], label=name)

plt.xlabel('Time (s)')
plt.ylabel('Control Signal')
plt.title('Perbandingan Control Signal vs Time')
plt.legend()
plt.grid()

plt.show()
```

**PLOT 3 — Subplot Sensor (SEMUA EKSPERIMEN)**

```python
fig, axs = plt.subplots(len(data), 1, figsize=(6, 10))

for i, (name, df) in enumerate(data.items()):
    axs[i].plot(df['time_s'], df['d_left'], label='Left')
    axs[i].plot(df['time_s'], df['d_right'], label='Right')

    axs[i].set_title(name)
    axs[i].set_xlabel('Time (s)')
    axs[i].set_ylabel('Distance')
    axs[i].legend()
    axs[i].grid()

plt.tight_layout()
plt.show()
```
